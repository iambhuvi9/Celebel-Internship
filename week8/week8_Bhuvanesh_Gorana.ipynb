Build a **Single-Agent Smart Assistant** that can:

- Understand user queries
- Identify user intent

````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str):
    """Evaluate a mathematical expression."""

    try:
        result = eval(expression)
        return result

    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str):
    """Extract important keywords."""

    try:

        words = text.split()

        keywords = []

        for word in words:

            word = word.lower().strip(",.!?")

            if len(word) > 4 and word not in keywords:
                keywords.append(word)

        return keywords[:5]

    except Exception:

        return []
````

````python
# 🛠️ TOOL 3: Greeting Tool

def greeting():

    return "Hello! 👋 Welcome to the Single Agent Pipeline."
````

## 🤖 Implement Agent Logic Below

The Agent performs three steps:

1. Understand the user's query.
2. Decide which tool should solve it.
3. Execute the selected tool.

### Routing Rules

- If the query contains **calculate** or a mathematical expression → Calculator Tool
- If the query contains **keywords** → Keyword Tool
- If the query is a greeting → Greeting Tool
- Otherwise → General Response

````python
# 🤖 AGENT FUNCTION

import json
import re
from datetime import datetime


def agent(query: str):

    query_lower = query.lower().strip()

    greetings = [
        "hi",
        "hello",
        "hey",
        "good morning",
        "good evening"
    ]

    try:

        # Greeting
        if query_lower in greetings:

            return {
                "type": "greeting",
                "tool": "Greeting Tool",
                "result": greeting(),
                "time": str(datetime.now())
            }

        # Calculator
        elif "calculate" in query_lower:

            expression = query_lower.replace("calculate", "").strip()

            return {
                "type": "calculation",
                "tool": "Calculator Tool",
                "result": calculator(expression),
                "time": str(datetime.now())
            }

        # Direct Math
        elif re.fullmatch(r"[0-9+\-*/(). ]+", query):

            return {
                "type": "calculation",
                "tool": "Calculator Tool",
                "result": calculator(query),
                "time": str(datetime.now())
            }

        # Keyword Extraction
        elif "keyword" in query_lower:

            text = query.replace("Extract keywords from", "")
            text = text.replace("extract keywords from", "").strip()

            return {
                "type": "keywords",
                "tool": "Keyword Tool",
                "result": extract_keywords(text),
                "time": str(datetime.now())
            }

        # General
        else:

            return {
                "type": "general",
                "tool": "None",
                "result": f"You asked: {query}",
                "time": str(datetime.now())
            }

    except Exception as e:

        return {
            "type": "error",
            "result": str(e)
        }
````

````python
# 🤖 AGENT FUNCTION (Updated)

import json
import re
import random
from datetime import datetime

def agent(query: str):

    query_lower = query.lower().strip()

    greetings = [
        "hi",
        "hello",
        "hey",
        "good morning",
        "good evening",
        "good afternoon"
    ]

    # ------------------------------------------------------
    # Knowledge Base
    # ------------------------------------------------------

    knowledge_base = {

        "artificial intelligence":
        "Artificial Intelligence (AI) is the simulation of human intelligence in machines that can think, learn, reason, and solve problems.",

        "machine learning":
        "Machine Learning (ML) is a branch of AI that enables computers to learn from data and improve their performance without being explicitly programmed.",

        "deep learning":
        "Deep Learning is a subset of Machine Learning that uses multi-layer neural networks to learn complex patterns from large datasets.",

        "data science":
        "Data Science is an interdisciplinary field that combines statistics, programming, machine learning, and data visualization to extract useful insights from data.",

        "python":
        "Python is a high-level programming language widely used for AI, Data Science, Machine Learning, Web Development, and Automation.",

        "agentic ai":
        "Agentic AI refers to intelligent AI systems that can understand goals, make decisions, select appropriate tools, execute tasks, and adapt their behavior with minimal human intervention.",

        "chatgpt":
        "ChatGPT is an AI chatbot developed by OpenAI that uses Large Language Models (LLMs) to understand and generate human-like text.",

        "llm":
        "A Large Language Model (LLM) is an AI model trained on massive amounts of text data to understand and generate natural language.",

        "generative ai":
        "Generative AI is a type of Artificial Intelligence capable of generating text, images, code, audio, and videos based on user prompts.",

        "neural network":
        "A Neural Network is a computational model inspired by the human brain and forms the foundation of most Deep Learning systems."
    }

    try:

        # ======================================================
        # Greeting Tool
        # ======================================================

        if query_lower in greetings:

            return {
                "type": "greeting",
                "tool": "Greeting Tool",
                "result": greeting(),
                "time": str(datetime.now())
            }

        # ======================================================
        # Calculator Tool
        # ======================================================

        elif query_lower.startswith("calculate"):

            expression = query_lower.replace("calculate", "").strip()

            return {
                "type": "calculation",
                "tool": "Calculator Tool",
                "result": calculator(expression),
                "time": str(datetime.now())
            }

        # ======================================================
        # Direct Math
        # ======================================================

        elif re.fullmatch(r"[0-9+\-*/(). ]+", query):

            return {
                "type": "calculation",
                "tool": "Calculator Tool",
                "result": calculator(query),
                "time": str(datetime.now())
            }

        # ======================================================
        # Keyword Extraction
        # ======================================================

        elif "keyword" in query_lower:

            text = query.replace("Extract keywords from", "")
            text = text.replace("extract keywords from", "").strip()

            return {

                "type": "keywords",

                "tool": "Keyword Tool",

                "result": extract_keywords(text),

                "time": str(datetime.now())

            }

        # ======================================================
        # Date & Time Tool
        # ======================================================

        elif "time" in query_lower:

            return {

                "type": "time",

                "tool": "Clock Tool",

                "result": datetime.now().strftime("%I:%M:%S %p"),

                "time": str(datetime.now())

            }

        elif "date" in query_lower:

            return {

                "type": "date",

                "tool": "Calendar Tool",

                "result": datetime.now().strftime("%d-%m-%Y"),

                "time": str(datetime.now())

            }

        # ======================================================
        # Joke Tool
        # ======================================================

        elif "joke" in query_lower:

            jokes = [

                "Why do programmers prefer Python? Because it has fewer bugs to fix! 😂",

                "Why was the computer cold? Because it left its Windows open! 😄",

                "Why did the AI go to school? To improve its neural network! 🤖",

                "Debugging: Being the detective in a crime movie where you're also the murderer. 😅"

            ]

            return {

                "type": "fun",

                "tool": "Joke Tool",

                "result": random.choice(jokes),

                "time": str(datetime.now())

            }

        # ======================================================
        # Knowledge Base Search
        # ======================================================

        else:

            for topic, answer in knowledge_base.items():

                if topic in query_lower:

                    return {

                        "type": "general",

                        "tool": "Knowledge Base",

                        "result": answer,

                        "time": str(datetime.now())

                    }

        # ======================================================
        # Default Response
        # ======================================================

            return {

                "type": "general",

                "tool": "General Response",

                "result": "Sorry, I couldn't understand your request. Please try asking a calculation, keyword extraction, greeting, date/time, joke, or a question about AI, Python, Machine Learning, or Data Science.",

                "time": str(datetime.now())

            }

    except Exception as e:

        return {

            "type": "error",

            "tool": "Agent",

            "result": str(e),

            "time": str(datetime.now())

        }
````

````python
# 🎯 Interactive Mode

print("="*60)
print("🤖 Welcome to the Single Agent Smart Assistant")
print("Type 'exit' anytime to stop.")
print("="*60)

while True:

    user_input = input("\n👤 You: ").strip()

    if user_input.lower() == "exit":

        print("\n👋 Thank you for using the Smart Assistant.")
        break

    response = agent(user_input)

    print("\n🤖 Agent Response")

    print(json.dumps(response, indent=4))
````
